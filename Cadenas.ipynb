## EJERCICIOS

Cual es el valor de la variable A después de ejecutar el siguiente código?

```python
# Escribe abajo tu código y presiona Shift+Enter para ejecutarlo

A = "1"
print(A)
# La variable "A" almacena el 1 como una cadena de texto, al estar entre comillas no lo almacena como entero
```

¿Cual es el valor de la variable B después de ejecutar el siguiente código?

```python
# Escribe abajo tu código y presiona Shift+Enter para ejecutarlo

B = "2"
print(B)
# La variable "B" almacena el 2 como una cadena de texto, al estar entre comillas no lo almacena como entero
```

¿Cual es el valor de la variable C después de ejecutar el siguiente código?

```python
# Escribe abajo tu código y presiona Shift+Enter para ejecutarlo

C = A + B
print(C)
# El valor termina siendo 12
# Las variables A y B al tener almacenados dos cadenas de texto, la suma de estas termina siendo la concatenación de ambos numeros.
```

En la variable D haz uso del slicing para imprimir solo los primeros tres elementos:

```python
# Escribe abajo tu código y presiona Shift+Enter para ejecutarlo

D = "ABCDEFG"
print(D[0:3])
#El slicing [0:3] nos da los elementos de la cadena que tienen indice del 0 al 2 que serian 'ABC'
```

Usa un valor de stride de 2 para imprimir cada segundo elemento de la cadena E:

```python
# Escribe abajo tu código y presiona Shift+Enter para ejecutarlo

E = 'clocrkr1e1c1t'
E[::2]
#Usando el stride con valor 2, se seleccionara cada segundo caracter de la cadena en este caso seria 'correct'
```

Imprime una barra invertida:

```python
# Escribe abajo tu código y presiona Shift+Enter para ejecutarlo
print("\\")
#Para imprimir correctamente la barra invertida se puede hacer poniendo dos veces \ para que se imprima correctamente
```

Convierte el mayúsculas el contenido de la variable F:

```python
# Escribe abajo tu código y presiona Shift+Enter para ejecutarlo

F = "You are wrong"
A = F.upper()
print(A)
#Al usar el metodo upper se transforma a mayusculas la cadena en la que se esté usando
```

En la variable G, encuentra el primer índice de la sub-cadena snow:

```python
# Escribe abajo tu código y presiona Shift+Enter para ejecutarlo

G = "Mary had a little lamb Little lamb, little lamb Mary had a little lamb \
Its fleece was white as snow And everywhere that Mary went Mary went, Mary went \
Everywhere that Mary went The lamb was sure to go"
G.find("snow")
#Usando el metodo "find" podemos encontrar en que posición se encuentra la sub-cadena
```

En la variable G, reemplaza la sub-cadena Mary con Bob:

```python
# Escribe abajo tu código y presiona Shift+Enter para ejecutarlo

G.replace('Mary', 'Bob')
#Se remplazan todos los mary que encuentre por Bob con replace
```
