## EJERCICIOS

Considérese la siguiente tupla:

```python
# Tupla de ejemplo

genres_tuple = ("pop", "rock", "soul", "hard rock", "soft rock", \
                "R&B", "progressive rock", "disco")
genres_tuple
```

Encontrar la longitud de la tupla, <code>genres_tuple</code>:

```python
# Escribe abajo tu código y presiona Shift+Enter para ejecutarlo
len(genres_tuple)
# Con len tenemos el tamaño de la tupla
```

Accede al elemento en el índice 3:

```python
# Escribe abajo tu código y presiona Shift+Enter para ejecutarlo
genres_tuple[3]
# Con [] y poniendo el numero del indice encontramos el elemento
```

Utiliza el slicing para obtener los índices 3, 4 y 5:

```python
# Escribe abajo tu código y presiona Shift+Enter para ejecutarlo
genres_tuple[3:6]
# Poniendo entre [] el valor del indice inicial y final nos da los elementos de los valores de adentro
```

Encuentra los primeros dos elementos de la tupla <code>genres_tuple</code>:

```python
# Escribe abajo tu código y presiona Shift+Enter para ejecutarlo
genres_tuple[0:2]
```

Encuentra el primer índice de <code>"disco"</code>:

```python
# Escribe abajo tu código y presiona Shift+Enter para ejecutarlo
genres_tuple.index("disco")
# Con index y poniendo el valor que buscamos nos da el indice
```

Genera una lista ordenada desde la tupla <code>C_tuple=(-5, 1, -3)</code>:

```python
# Escribe abajo tu código y presiona Shift+Enter para ejecutarlo
C_tuple=(-5, 1, -3)
TupleSorted = sorted(C_tuple)
TupleSorted
# Creamos la tupla y una variable en la que se guarde organizada, sorted nos ayuda a ordenar los valores
```
